Translate the 'sent' column in the TSV file "/content/MADAR.corpus.MSA.tsv" to the Kuwaiti dialect using OpenAI's GPT-4o model.

```python
import pandas as pd

# Read the TSV file into a pandas DataFrame
df_msa = pd.read_csv('/content/MADAR.corpus.MSA.tsv', sep='\t')

# Display the first few rows of the DataFrame
display(df_msa.head())

# Display the information about the DataFrame
display(df_msa.info())
```

```python
import os
from google.colab import userdata

# Read the OpenAI API key from Colab secrets
openai_api_key = userdata.get('OPENAI_API_KEY')

# Set the OpenAI API key as an environment variable
os.environ["OPENAI_API_KEY"] = openai_api_key
```

Translation function



```python
from openai import OpenAI

def translate_to_kuwaiti(sentence):

    try:
        client = OpenAI()
        response = client.chat.completions.create(
            model="gpt-4o",
            messages=[
                {"role": "system", "content": "You are a helpful assistant that translates text to the Kuwaiti dialect."},
                {"role": "user", "content": f"Translate the following sentence to Kuwaiti dialect: {sentence}"}
            ],
            temperature=0.7,
            max_tokens=150,
            top_p=1.0,
            frequency_penalty=0.0,
            presence_penalty=0.0
        )
        translated_text = response.choices[0].message.content
        return translated_text
    except Exception as e:
        print(f"An error occurred during translation: {e}")
        return None
```

Translate sentences


```python
df_msa['sent_kuwaiti'] = df_msa['sent'].apply(translate_to_kuwaiti)
display(df_msa.head())
```

```python
df_KW = df_msa.copy()
display(df_KW.head())
```

Save Kuwaiti dataset as CSV


```python
df_KW.to_csv('MADAR.corpus.KW.csv', sep=',', index=False)
print('df_KW saved as MADAR.corpus.KW.csv')
```
